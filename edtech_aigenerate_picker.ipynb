# edtech aigenerate picker


My app helps educational technology companies decide which frontier Claude model to use for generating educational passages by using an LLM judge and word-count analysis to compare each model’s length accuracy, clarity, and factual correctness across subjects, task types, and difficulty levels.

The data came in two seperate data sets:
1. Prompt -- identified by the prompt ID, it specifies the domain, task type, topic, and prompt language of each prompt
2. Responses or "ntl_extra_passages_claude" -- identified by a unique id, this table holds the prompt id, model, text response, words, domain, task type, and topic.

I will begin by only using the Responses table as that holds the results, but if I need more information about the prompts, I will join the tables later on.

```python
def average_words_by_model(df):

    # Calculate average number of words for each model
    average_words = df.groupby("model")["n_words"].mean()

    # Sort results from lowest to highest
    average_words = average_words.sort_values()

    return average_words

# Display results from shortest to longest
print(average_words)
```
